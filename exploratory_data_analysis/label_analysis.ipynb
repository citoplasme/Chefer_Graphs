```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import plotnine as p9
```

```python
def get_label_information(train, validation, test, annotation_theme = (1.5, 20, 0)):
  train['split'] = 'Training'
  validation['split'] = 'Validation'
  test['split'] = 'Testing'
  df = pd.concat([train, validation, test])

  plot_df = df[['split', 'label']].groupby('split', as_index = False).value_counts(normalize = True)
  plot_df['proportion'] = plot_df['proportion'].round(4) * 100
  plot_df['proportion_label'] = plot_df['proportion'].round(2)
  plot_df['label'] = plot_df['label'].astype(int)
  plot_df['label'] = pd.Categorical(plot_df['label'], categories = plot_df['label'].unique().sort())
  plot_df['split'] = pd.Categorical(plot_df['split'], categories = ['Training', 'Validation', 'Testing'])

  display(plot_df)

  return p9.ggplot(
    plot_df,
    p9.aes(x = 'label', y = 'proportion', fill = 'split')
  ) + \
    p9.geom_bar(stat = 'identity', position = p9.position_dodge2(), colour = '#000000') + \
    p9.geom_text(p9.aes(y = f'proportion + {annotation_theme[0]}', label = 'proportion_label'), position = p9.position_dodge2(width = .9), size = annotation_theme[1], angle = annotation_theme[2]) + \
    p9.ylab('Proportion (%)') + \
    p9.xlab('Label') + \
    p9.labs(fill = 'Split') + \
    p9.theme_bw() + \
    p9.scale_fill_manual(values = ['#ef8a62', '#67a9cf', '#ffd760', '#f7f7f7']) + \
    p9.theme(
      legend_position = 'top',
      legend_background = p9.element_rect(color = '#000000', fill = '#ebebeb'),
      legend_box_margin = 0,
      figure_size = (12, 8),
      text = p9.element_text(size = 25, family = 'serif'),
      axis_text = p9.element_text(size = 22),
      axis_title = p9.element_text(weight = 'bold'),
      legend_title = p9.element_text(weight = 'bold'),
    )
```

## IMDb

```python
training_df = pd.read_csv('../data/with_validation_splits/IMDb/train.csv')
validation_df = pd.read_csv('../data/with_validation_splits/IMDb/validation.csv')
testing_df = pd.read_csv('../data/with_validation_splits/IMDb/test.csv')
```

```python
training_df.shape[0], validation_df.shape[0], testing_df.shape[0]
```

```python
get_label_information(
  training_df,
  validation_df,
  testing_df,
)
```
